```python
data = {"name":["nick", "judy", "alex"], "age" :[10,15,20]}
data
df = pd.DataFrame(data)
df 
```

```python
## 기존 열을 활용하여 새로운 열을 쉽게 만들 수 있음.
```

```python
df['age_plus'] =df['age'] +1
df['age_squared'] = df['age'] * df['age']
df
```

```python
## join 가능 
```

```python
df2 = pd.DataFrame({ "name": ["nick", "judy", "alex"]
                     ,"height" : [180, 165, 175]
                     , "gender" : ["M", "F", "M"]})
joined = df.set_index("name").join(df2.set_index("name"))
joined
```

```python
## group by 가능
```

```python
df_mean =joined.groupby("gender").mean()
df_mean
```
